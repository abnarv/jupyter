```python
import pandas as pd
import numpy as np 
from sklearn import linear_model
%matplotlib inline 
import matplotlib.pyplot as plt

df = pd.read_csv('C:\\Users\\Abha Mandar Narvekar\\Documents\\vgsales.csv')
df
#errors: double up the backslashes in the file path 
```

```python

plt.scatter(df.Area, df.Price)
plt.xlabel('Area in Sq ft')
plt.ylabel('Price in USD')
plt.show()
```

```python
reg = linear_model.LinearRegression() #creating an object 
reg.fit(df[['Area']],df.Price) #reg.fit([2D array], 1D array)
```

```python
reg.predict([[3500]])
```

```python
reg.coef_ #slope of the line 
```

```python
reg.intercept_ #y intercept of the line 
```

```python
# 
d = pd.read_csv("C:\\Users\\Abha Mandar Narvekar\\Documents\\areas.csv")
d
```

```python
p = reg.predict(d[['Area']])
d['Prices'] = p #creates a new column which will contain all the predicted prices
d.to_csv("C:\\Users\\Abha Mandar Narvekar\\Documents\\areas.csv")
```
