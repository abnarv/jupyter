```python
#CSTRs in series
import numpy as np
import matplotlib.pyplot as plt

n = np.array([1, 2, 3, 4, 5, 6, 7])
k_tau = np.array([0.01, 0.05, 0.1, 0.2, 0.25, 0.3, 0.35])

for kt in k_tau:
    X = 1 - 1 / ((1 + kt) ** n)   # vectorized over n
    plt.plot(n, X, marker='o', label=f'kτ = {kt}')

plt.xlabel('Number of CSTRs in series (n)')
plt.ylabel('Conversion (X)')
plt.title('Conversion vs Number of CSTRs')
plt.legend()
plt.grid()
plt.show()
```
